## Preparacion de los datos

EI proceso consiste en transformar las variables categóricas y la
variable objetivo a formatos numéricos que los algoritmos de ML puedan interpretar.

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler

# ============ PREPARACIÓN ============
# Identificar tipos de columnas
categorical_features = X_train.select_dtypes(include=['object']).columns.tolist()
numerical_features = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("---Preparacion---")
print(f"Numéricas ({len(numerical_features)}): {numerical_features}")
print(f"Categóricas ({len(categorical_features)}): {categorical_features}")

# ============ CREAR PIPELINE DE PREPROCESAMIENTO ============

# One-Hot Encoding con drop_first para consistencia con tu approach actual
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_features)
    ],
    remainder='drop'  # Por seguridad
)

# Label Encoder para la variable objetivo
label_encoder = LabelEncoder()

# ============ FIT DEL PREPROCESAMIENTO (solo con train) ============
X_train_processed = preprocessor.fit_transform(X_train)
y_train_processed = label_encoder.fit_transform(y_train)

# Obtener nombres de columnas procesadas para referencia
feature_names = (numerical_features +
                 list(preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)))

print(f"\n--- Preprocesamiento X ---")
print(f"Shape original: {X_train.shape}")
print(f"Shape después de encoding: {X_train_processed.shape}")
print(f"Nuevas columnas ({len(feature_names)}): {feature_names}")

print(f"\n--- Preprocesamiento Y ---")
print(f"Clases: {label_encoder.classes_} → {label_encoder.transform(label_encoder.classes_)}")

# ============ TRANSFORMAR TEST (sin fit) ============
X_test_processed = preprocessor.transform(X_test)
y_test_processed = label_encoder.transform(y_test)

print(f"\n--- Test procesado ---")
print(f"Shape: {X_test_processed.shape}")
```

#### Creación de función de inferencia

```python
def predict_student_risk(student_data: dict) -> dict:
    """
    Recibe datos crudos de un estudiante (dict o DataFrame de 1 fila)
    y devuelve (clase, probabilidad).

    Args:
        student_data: dict con las mismas columnas que X_train original

    Returns:
        dict con 'class', 'probability', 'class_label'
    """
    # Cargar artefactos (en producción se cargan una sola vez al iniciar)
    preprocessor = joblib.load('model_artifacts/preprocessor.joblib')
    label_encoder = joblib.load('model_artifacts/label_encoder.joblib')
    model = joblib.load('model_artifacts/logreg_optimized.joblib')
    metadata = joblib.load('model_artifacts/metadata.joblib')

    # Convertir a DataFrame si es dict
    if isinstance(student_data, dict):
        student_df = pd.DataFrame([student_data])
    else:
        student_df = student_data.copy()

    # Asegurar que tenga las columnas correctas
    expected_cols = metadata['numerical_features'] + metadata['categorical_features']
    missing = set(expected_cols) - set(student_df.columns)
    if missing:
        raise ValueError(f"Faltan columnas: {missing}")

    # Reordenar columnas para consistencia
    student_df = student_df[expected_cols]

    # Preprocesar (solo transform, nunca fit)
    X_processed = preprocessor.transform(student_df)

    # Predecir
    proba = model.predict_proba(X_processed)[0]  # [prob_clase_0, prob_clase_1]
    pred_class = model.predict(X_processed)[0]

    return {
        'class': int(pred_class),
        'class_label': label_encoder.inverse_transform([pred_class])[0],
        'probability_risk': float(proba[1]),      # Probabilidad de clase 1 (risk)
        'probability_not_risk': float(proba[0]),  # Probabilidad de clase 0
        'risk_percentage': f"{proba[1]*100:.2f}%"
    }
```

#### 2. Validación cruzada del modelo base

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_accuracy = cross_val_score(logreg_base, X_train_processed, y_train_processed,
                               cv=cv, scoring='accuracy')
cv_auc = cross_val_score(logreg_base, X_train_processed, y_train_processed,
                         cv=cv, scoring='roc_auc')
cv_f1 = cross_val_score(logreg_base, X_train_processed, y_train_processed,
                         cv=cv, scoring='f1_weighted')

print("=== VALIDACIÓN CRUZADA (5-Fold) ===\n")
print(f"Accuracy CV: {cv_accuracy.mean():.4f} (+/- {cv_accuracy.std():.4f})")
print(f"AUC CV: {cv_auc.mean():.4f} (+/- {cv_auc.std():.4f})")
print(f"F1-Score CV: {cv_f1.mean():.4f} (+/- {cv_f1.std():.4f})")
print(f"\nScores individuales Accuracy: {cv_accuracy.round(4)}")
print(f"Scores individuales AUC: {cv_auc.round(4)}")
```

#### 3. Optimización de hiperparámetros

```python
# Definir parámetros a probar
param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100],        # Regularización inversa
    'penalty': ['l1', 'l2'],                      # Tipo de regularización
    'solver': ['liblinear', 'saga'],              # Solvers compatibles
    'class_weight': [None, 'balanced'],           # Manejo de desbalance
    'max_iter': [1000, 2000]                      # Iteraciones suficientes
}

print("=== OPTIMIZACIÓN DE HIPERPARÁMETROS ===\n")
print(f"Total de combinaciones: {len(param_grid['C']) * len(param_grid['penalty']) * len(param_grid['solver']) * len(param_grid['class_weight']) * len(param_grid['max_iter'])}")
print("Esto puede tomar unos minutos...\n")

# Grid Search con validación cruzada
grid_search = GridSearchCV(
    LogisticRegression(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train_processed, y_train_processed)

print("\n=== MEJORES HIPERPARÁMETROS ===\n")
for param, value in grid_search.best_params_.items():
    print(f"{param}: {value}")

print(f"\nMejor score CV: {grid_search.best_score_:.4f}")
```

```python
# ============ RECONSTRUIR X_test_processed COMO DATAFRAME ============
feature_names = preprocessor.get_feature_names_out()

# Crear DataFrame con nombres de features
X_test_df = pd.DataFrame(X_test_processed, columns=feature_names, index=X_test.index)

# ============ ANÁLISIS DE ERRORES ============
errores = X_test_df.copy()
errores['Real'] = y_test_processed
errores['Real_Texto'] = label_encoder.inverse_transform(y_test_processed)
errores['Predicho'] = y_pred_opt
errores['Predicho_Texto'] = label_encoder.inverse_transform(y_pred_opt)
errores['Probabilidad_riesgo'] = y_pred_proba_opt

# Identificar errores
fp = errores[(errores['Real'] == 0) & (errores['Predicho'] == 1)]  # not_risk predicho como risk
fn = errores[(errores['Real'] == 1) & (errores['Predicho'] == 0)]  # risk predicho como not_risk

print("=== ANÁLISIS DE ERRORES ===\n")
print(f"Total de registros en prueba: {len(errores)}")
print(f"Aciertos: {len(errores) - len(fp) - len(fn)} ({((len(errores) - len(fp) - len(fn))/len(errores)*100):.1f}%)")
print(f"Falsos Positivos ({label_encoder.classes_[0]} → {label_encoder.classes_[1]}): {len(fp)} ({len(fp)/len(errores)*100:.1f}%)")
print(f"Falsos Negativos ({label_encoder.classes_[1]} → {label_encoder.classes_[0]}): {len(fn)} ({len(fn)/len(errores)*100:.1f}%)")

# Mostrar ejemplos de errores
if len(fp) > 0:
    print("\n" + "="*60)
    print(f"EJEMPLO DE FALSO POSITIVO ({label_encoder.classes_[0]} predicho como {label_encoder.classes_[1]}):")
    print("="*60)
    fp_sample = fp.iloc[0]
    # Mostrar solo las primeras 5 columnas de features (excluyendo las columnas agregadas)
    feature_cols = list(X_test_df.columns[:5])
    for col in feature_cols:
        print(f"{col}: {fp_sample[col]}")
    print(f"Probabilidad de riesgo: {fp_sample['Probabilidad_riesgo']:.3f}")

if len(fn) > 0:
    print("\n" + "="*60)
    print(f"EJEMPLO DE FALSO NEGATIVO ({label_encoder.classes_[1]} predicho como {label_encoder.classes_[0]}):")
    print("="*60)
    fn_sample = fn.iloc[0]
    feature_cols = list(X_test_df.columns[:5])
    for col in feature_cols:
        print(f"{col}: {fn_sample[col]}")
    print(f"Probabilidad de riesgo: {fn_sample['Probabilidad_riesgo']:.3f}")

# Análisis de probabilidades
if len(fp) > 0:
    print(f"\nLos falsos positivos tienen probabilidad de riesgo promedio: {fp['Probabilidad_riesgo'].mean():.3f}")
if len(fn) > 0:
    print(f"Los falsos negativos tienen probabilidad de riesgo promedio: {fn['Probabilidad_riesgo'].mean():.3f}")
```

#### 4. Curvas de Aprendizaje

```python
from sklearn.model_selection import learning_curve

train_sizes, train_scores, test_scores = learning_curve(
    logreg_optimized, X_train_processed, y_train_processed,
    cv=5, scoring='roc_auc',
    train_sizes=np.linspace(0.1, 1.0, 10),
    n_jobs=-1
)

train_mean = train_scores.mean(axis=1)
test_mean = test_scores.mean(axis=1)

plt.figure(figsize=(10, 6))
plt.plot(train_sizes, train_mean, label='Entrenamiento', marker='o', linewidth=2)
plt.plot(train_sizes, test_mean, label='Validación Cruzada', marker='s', linewidth=2)
plt.xlabel('Tamaño del conjunto de entrenamiento', fontsize=12)
plt.ylabel('AUC', fontsize=12)
plt.title('Curvas de Aprendizaje - Logistic Regression', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Diagnóstico
final_gap = train_mean[-1] - test_mean[-1]
print("\n=== DIAGNÓSTICO DE APRENDIZAJE ===\n")
print(f"Brecha final train-val: {final_gap:.4f}")

if final_gap > 0.1:
    print("Posible OVERFITTING: El modelo no generaliza bien")
elif final_gap > 0.05:
    print("Cierta señal de overfitting")
elif final_gap < 0:
    print("Posible UNDERFITTING: El modelo es demasiado simple")
else:
    print("BUEN AJUSTE: El modelo generaliza correctamente")
```
